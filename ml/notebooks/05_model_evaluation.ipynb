# ShopFlow — Model Evaluation & Comparison

Compares all trained models and displays evaluation metrics. Ensures no fake or misleading performance (VISION.md §7).

```python
import sys
sys.path.insert(0, ".")

import mlflow
import pandas as pd
from ml.evaluation.metrics import regression_metrics, classification_metrics

mlflow.set_tracking_uri("http://localhost:5000")

# List recent runs
runs = mlflow.search_runs(experiment_ids=["0"])
print("MLflow runs:", len(runs))
print(runs[['run_id', 'status', 'start_time']].head(10).to_string(index=False))
```

```python
# Example: evaluate delivery ETA predictions
import numpy as np

y_true = np.array([5.2, 8.1, 3.0, 12.0, 6.5])
y_pred = np.array([5.0, 7.5, 3.5, 11.0, 7.0])

metrics = regression_metrics(y_true, y_pred)
print("Delivery ETA regression metrics:")
for k, v in metrics.items():
    print(f"  {k}: {v:.4f}")
```
