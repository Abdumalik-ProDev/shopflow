# ShopFlow — Recommendation Model

Trains recommendation models:
- Popularity-based baseline
- Collaborative filtering / ALS / LightFM (placeholders)

Logs metrics and top products to MLflow. Per VISION.md §4: Product Recommendation — benchmark popularity, collaborative filtering, ALS, LightFM, neural approaches.

```python
import sys
sys.path.insert(0, ".")

import mlflow
from ml.pipelines.recommendation_pipeline import train_recommendation_model

mlflow.set_tracking_uri("http://localhost:5000")

result = train_recommendation_model()
print("\n=== Recommendation Training Result ===")
print(result)

print("\nTop recommended products:")
for p in result.get("top_products", []):
    print(f"  {p['product_id']} | {p['product_category_name']} | revenue={p['revenue']:.2f} | units={p['units']}")
```
