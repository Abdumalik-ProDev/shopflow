# ShopFlow — Feature Engineering

Builds model-ready feature frames from the Olist dataset.

- `build_order_features()` → one row per order with items, payments, reviews
- `build_customer_features()` → RFM per customer
- `build_product_features()` → aggregated product metrics

Outputs are written to `ml/data/cleared/` for consumption by training pipelines.

```python
import sys
sys.path.insert(0, ".")

from ml.features import (
    build_order_features,
    build_customer_features,
    build_product_features,
    OLIST_DIR,
)
import pandas as pd
from pathlib import Path

CLEARED_DIR = Path("data/cleared")
CLEARED_DIR.mkdir(parents=True, exist_ok=True)

print("Reading raw data from:", OLIST_DIR)
```

```python
# Build order-level features (delivery ETA target)
order_df = build_order_features()
print("Order features shape:", order_df.shape)
print(order_df.head())
order_df.to_csv(CLEARED_DIR / "order_features.csv", index=False)
print("Saved → data/cleared/order_features.csv")
```

```python
# Build customer-level RFM features
customer_df = build_customer_features()
print("Customer features shape:", customer_df.shape)
print(customer_df.head())
customer_df.to_csv(CLEARED_DIR / "customer_features.csv", index=False)
print("Saved → data/cleared/customer_features.csv")
```

```python
# Build product-level features
product_df = build_product_features()
print("Product features shape:", product_df.shape)
print(product_df.head())
product_df.to_csv(CLEARED_DIR / "product_features.csv", index=False)
print("Saved → data/cleared/product_features.csv")
```

```python
# Summary of cleared files
for f in sorted(CLEARED_DIR.glob("*.csv")):
    df = pd.read_csv(f)
    print(f"{f.name:35s} {df.shape[0]:>6,} rows x {df.shape[1]:>2} cols")
```
